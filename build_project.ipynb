```python
import subprocess
import os

# Change to build directory
os.chdir('/Users/maxim.podkolzin/dev/cpp/pesdb/build')

# Run cmake build
print('Building project...')
print('=' * 80)
result = subprocess.run(['cmake', '--build', '.'], capture_output=True, text=True)

print('STDOUT:')
print(result.stdout)

if result.stderr:
    print('\nSTDERR:')
    print(result.stderr)

print('\n' + '=' * 80)
if result.returncode == 0:
    print('BUILD SUCCESSFUL')
else:
    print(f'BUILD FAILED (return code: {result.returncode})')

print('=' * 80)
```
